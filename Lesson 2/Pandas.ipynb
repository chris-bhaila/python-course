## Handling duplicated values

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'SN': [1, 2, 3, 4, 2, 6, 4],
    'Age': [32, 5, 42, 34, 5, 31, 34],
    'Roll': ['A01', 'B07', 'C03', 'df4', 'B07', 'F08', 'df4'],
    'Employee ID': [100234.0, 102345.0, 105678.0, 23242.0, 102345.0, np.nan, 23242.0],
    'Name': ['mmd', 'Jane Smith', 'David Brown', 'miraj', 'Jane Smith', 'Anna', 'miraj']
})

df
```

```python
df[df.duplicated()]
```

```python
df.drop_duplicates(inplace=True)
df
```

```python
~df.duplicated() #Where there are no duplicates, it will return True, and where there are duplicates, it will return False.
```
