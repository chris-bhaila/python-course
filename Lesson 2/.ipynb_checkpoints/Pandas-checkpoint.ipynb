```python
import pandas as pd

# Creating series with list
l = [1,2,3,4,5]
s = pd.Series(l)
s
```

```python
s = pd.Series(l, index=['a','b','c','d','e'])
s
```

```python
import numpy as np

a = np.array([10,50,20])
pd.Series(a)
```

```python
#Creating Series with Dictionary
d={'name':'miraj','roll':10,'age':15}
s=pd.Series(d)
s
```
