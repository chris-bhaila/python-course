# 1. Try-Except Blocks

## 1.1 What is an Exception?

An **exception** is an error that happens **while** your program is running. Without handling, an exception will **crash** the program.

Common examples:

| Exception | When it happens |
|-----------|-----------------|
| `ZeroDivisionError` | Dividing by zero |
| `ValueError` | Wrong value type (e.g., `int("abc")`) |
| `TypeError` | Wrong type of operand (e.g., `"5" + 5`) |
| `NameError` | Variable not defined |
| `IndexError` | List index out of range |
| `KeyError` | Missing dictionary key |
| `FileNotFoundError` | File doesn't exist |
| `AttributeError` | Object has no such attribute/method |

## 1.2 Syntax of `try-except`

```python
try:
    # code that MIGHT raise an exception
except SomeError:
    # code that runs IF that exception occurs
```

Python will:
1. Try to run the code in the `try` block.
2. If an exception occurs, jump to the matching `except` block.
3. If no exception occurs, skip the `except` block.

# 2. Catching Specific Exceptions

It's a **best practice** to catch the **specific** exception you expect, not a generic one. This way you don't accidentally hide unrelated bugs.

### Multiple `except` blocks

You can have several `except` blocks - Python will use the first one that matches.

```python
try:
    ...
except ValueError:
    ...
except ZeroDivisionError:
    ...
```

### Multiple exceptions in one `except`

```python
try:
    ...
except (ValueError, TypeError):
    ...
```

````python
# Multiple except blocks - each handles a different error
def divide(a, b):
    try:
        a = int(a)
        b = int(b)
        return a / b
    except ValueError:
        return "Please provide valid numbers."
    except ZeroDivisionError:
        return "Cannot divide by zero."

print(divide(10, 2))          # 5.0
print(divide("abc", 5))       # ValueError
print(divide(10, 0))          # ZeroDivisionError
````

````python
# Multiple exception types in one except
def first_item(value):
    try:
        return value[0] # Type
    except (IndexError, KeyError, TypeError) as e:
        return f"Could not get first item: {type(e).__name__}: {e}"

print(first_item([1, 2, 3]))     # works
print(first_item([]))            # IndexError
print(first_item({}))            # KeyError (0 not in dict)
print(first_item(1))          # TypeError (None not subscriptable)
````

````python
# Accessing exception details with 'as e'
data = {"name": "Alice", "age": 25}

try:
    print(data["email"])
except KeyError as e:
    print("Missing key  :", e)            # the key that was missing
    print("Error type   :", type(e).__name__)
    print("Args         :", e.args)
````

````python
# Order matters - more specific exceptions FIRST
# Otherwise the generic 'Exception' will catch them all

def parse_number(text):
    try:
        return int(text)
    except ValueError:                      # specific FIRST
        return "Not a number"
    except Exception:                       # generic LAST (catches anything else)
        return "Unknown error"

print(parse_number("42"))
print(parse_number("abc"))
````

# 3. Using `else` and `finally`

The full structure of an exception handler looks like this:

```python
try:
    # code that might fail
except SomeError:
    # runs IF an exception occurs
else:
    # runs ONLY if NO exception occurred
finally:
    # ALWAYS runs (success or failure)
```

| Block | When it runs |
|-------|--------------|
| `try` | First - the protected code |
| `except` | Only if a matching exception is raised |
| `else` | Only when `try` succeeded without any exception |
| `finally` | **Always**, even after a `return` or unhandled error |

**Common uses:**
- `else` -> code that should run only if the operation was successful (e.g., process the loaded data).
- `finally` -> cleanup (closing files, releasing locks, disconnecting DB).

````python
# Real-world example: file handling with finally for cleanup
file = None
try:
    file = open("missing_file.txt", "r")
    data = file.read()
except FileNotFoundError:
    print("File not found - using default content.")
    data = ""
else:
    print("File read successfully.")
finally:
    if file is not None:
        file.close()
    print("Cleanup done.")
````

````python
# 'finally' even runs after a return statement
def test():
    try:
        print("In try")
        return "From try"
    finally:
        print("In finally (still runs before return!)")

print("Returned:", test())
````

## Quick Summary

| Concept | One-line takeaway |
|---------|-------------------|
| Exception | A runtime error that can be caught and handled |
| `try / except` | Wrap risky code so it doesn't crash the program |
| Specific `except` | Always prefer specific errors over generic `Exception` |
| Multiple `except` | First match wins - put specific ones FIRST |
| `else` | Runs only if the `try` block succeeded |
| `finally` | Runs ALWAYS - perfect for cleanup |
| `raise` | Trigger an exception yourself |
