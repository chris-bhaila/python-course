# 1. Classes & Objects

## 1.1 What is OOP?

**Object-Oriented Programming (OOP)** is a way of organizing code by grouping related data and the functions that work on that data into a single unit called an **object**.

### Why OOP?
- **Reusability** - write once, reuse via inheritance
- **Modularity** - keep related data & behavior together
- **Real-world modeling** - represent things like `Car`, `Student`, `BankAccount`
- **Maintainability** - easier to update & extend large programs

## 1.2 Class vs Object

| Term | Meaning | Example |
|------|---------|---------|
| **Class** | A blueprint / template | `Car` (the design) |
| **Object** | An instance of a class | `my_car = Car()` (an actual car) |

### Syntax

```python
class ClassName:
    # class body
    pass

obj = ClassName()        # create an object
```

````python
# A very simple class
class Dog:
    pass

# Creating objects (instances) of the class
dog1 = Dog()
dog2 = Dog()

print("dog1 :", dog1)
print("dog2 :", dog2)
print("type :", type(dog1))
````

````python
# A class with attributes and a method
class Dog:
    # method = function defined inside a class
    def bark(self):
        print("Woof! Woof!")

# Create an object and call its method
d = Dog()
d.bark() #bark(d) self is passed automatically

# Add attributes on the fly
d.name = "Buddy"
d.age = 3
print(f"{d.name} is {d.age} years old")
````

# 2. The `__init__()` Method & `self` Keyword

## 2.1 `__init__()` - The Constructor

`__init__()` is a **special method** (called a constructor) that Python runs **automatically** whenever you create a new object. It is used to initialize the object's attributes.

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
```

## 2.2 The `self` Keyword

`self` refers to **the current object**. It is the first parameter of every instance method and lets you access the object's own attributes and methods.

| Term | Meaning |
|------|---------|
| `self.name` | Attribute belonging to the object |
| `self.method()` | Method called on the object |
| `self` (parameter) | Auto-passed by Python when you call `obj.method()` |

Note: You don't pass `self` manually. `p.greet()` is automatically translated by Python into `Person.greet(p)`.

````python
# Class with __init__ constructor
class Person:
    def __init__(self, name, age):
        # 'self' refers to the new object being created
        self.name = name        # attribute
        self.age = age

    def greet(self):
        print(f"Hi, I'm {self.name} and I'm {self.age} years old.")

# Create objects - arguments go to __init__
p1 = Person("Alice", 25)
p2 = Person("Bob", 30)

p1.greet()
p2.greet()
````

````python
# Accessing and modifying attributes from outside
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

s = Student("Miraj", 85)
print("Name  :", s.name)
print("Marks :", s.marks)

# Modify the attribute
s.marks = 95
print("Updated Marks :", s.marks)
````

````python
# Default values in __init__
class Car:
    def __init__(self, brand, color="White", year=2024):
        self.brand = brand
        self.color = color
        self.year = year

    def describe(self):
        print(f"{self.year} {self.color} {self.brand}")

c1 = Car("Toyota")
c2 = Car("BMW", "Black")
c3 = Car("Tesla", "Red", 2025)

c1.describe()
c2.describe()
c3.describe()
````

# 3. Class Attributes vs Object (Instance) Attributes

| Attribute Type | Where Defined | Shared? | Access |
|----------------|---------------|---------|--------|
| **Class attribute** | Inside the class, **outside** any method | Yes - shared by ALL objects | `ClassName.attr` or `obj.attr` |
| **Object (instance) attribute** | Inside `__init__` using `self.x = ...` | No - unique to each object | `obj.attr` |

### Rule of Thumb
- Use a **class attribute** for data that is the same for every object (e.g., `species = 'Dog'`).
- Use an **instance attribute** for data that is unique to each object (e.g., `name`, `age`).

````python
# Class attribute vs Instance attribute
class Dog:
    species = "Canis familiaris"      # CLASS attribute - shared

    def __init__(self, name, age):
        self.name = name              # INSTANCE attribute - unique
        self.age = age

d1 = Dog("Buddy", 3)
d2 = Dog("Max", 5)


# Instance attributes are different
print(d1.name, d1.age)
print(d2.name, d2.age)

# Class attribute is the SAME for all objects
print(d1.species)
print(d2.species)
print(Dog.species)        # access directly via class
````

````python
class Dog1:
    def __init__(self, name, age, species="Canis familiaris"):
        self.name = name
        self.age = age
        self.species = species

d1 = Dog1("Buddy", 3)
print(d1.name, d1.age, d1.species)
````

````python
# Modifying class attribute affects all objects
class Dog:
    species = "Canis familiaris"

    def __init__(self, name):
        self.name = name

d1 = Dog("Buddy")
d2 = Dog("Max")

# Change at class level -> affects ALL instances
Dog.species = "Wolf descendant"
print("d1.species :", d1.species)
print("d2.species :", d2.species)

# Change on one object -> only that object gets a new INSTANCE attribute
d1.species = "Pug"
print("\nAfter modifying d1.species:")
print("d1.species :", d1.species)     # Pug (instance)
print(d1.__dict__)
print("d2.species :", d2.species)     # still Wolf descendant
print(d2.__dict__)
print("Dog.species:", Dog.species)    # still Wolf descendant
````

````python
# Common use - counting objects with a class attribute
class Employee:
    count = 0                          # class attribute

    def __init__(self, name):
        self.name = name
        Employee.count += 1            # update class attribute

e1 = Employee("Alice")
e2 = Employee("Bob")
e3 = Employee("Charlie")

print("Total employees :", Employee.count)
````

# 4. Inheritance

**Inheritance** lets a new class (**child**) reuse the attributes and methods of an existing class (**parent**).

```python
class Child(Parent):
    ...
```

### Why use Inheritance?
- **Code reuse** - don't rewrite shared logic
- **Hierarchy** - model `is-a` relationships (a Dog *is an* Animal)
- **Extensibility** - add new behavior on top of existing classes

### Types of Inheritance

| Type | Description | Example |
|------|-------------|---------|
| **Single** | One child, one parent | `Dog(Animal)` |
| **Multi-level** | Child inherits from a child of another class | `Puppy(Dog(Animal))` |
| **Multiple** | One child inherits from multiple parents | `Child(Father, Mother)` |
| **Hierarchical** | Multiple children from one parent | `Dog(Animal)`, `Cat(Animal)` |

## 4.1 Single Inheritance

One child class inherits from one parent class.

````python
# Single inheritance
class Animal:
    def __init__(self, name):
        self.name = name

    def eat(self):
        print(f"{self.name} is eating.")

class Dog(Animal):              # Dog inherits from Animal
    def bark(self):
        print(f"{self.name} says Woof!")

d = Dog("Buddy")
d.eat()                         # inherited from Animal
d.bark()                        # defined in Dog

# Check the relationship
print(isinstance(d, Dog))       # True
print(isinstance(d, Animal))    # True
print(issubclass(Dog, Animal))  # True
````

## 4.2 Multi-level Inheritance

A class inherits from a class that already inherits from another class. Creates a chain: `A -> B -> C`.

````python
# Multi-level inheritance
class Animal:
    def eat(self):
        print("Eating...")

class Dog(Animal):              # Dog -> Animal
    def bark(self):
        print("Barking...")

class Puppy(Dog):               # Puppy -> Dog -> Animal
    def weep(self):
        print("Weeping...")

p = Puppy()
p.eat()       # from Animal (grandparent)
p.bark()      # from Dog (parent)
p.weep()      # from Puppy (self)
````

## 4.3 Multiple Inheritance

One class inherits from **two or more** parent classes. Python uses **MRO (Method Resolution Order)** to decide which parent's method to call first.

Use `ClassName.__mro__` or `ClassName.mro()` to see the order.

````python
# Multiple inheritance
class Father:
    def skills(self):
        print("Father: Coding, Gardening")

class Mother:
    def skills(self):
        print("Mother: Cooking, Painting")

class Child(Father, Mother):    # inherits from BOTH
    def own_skill(self):
        print("Child: Gaming")

c = Child()
c.skills()                      # calls Father's (left-most priority)
c.own_skill()

# Method Resolution Order
print("\nMRO :", [cls.__name__ for cls in Child.__mro__])
````

## 4.4 The `super()` Function

`super()` lets you call methods of the parent class - especially useful inside `__init__` to reuse the parent's initialization.

````python
# Using super() to call parent's __init__
class Animal:
    def __init__(self, name):
        self.name = name
        print(f"Animal init: {self.name}")

class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)        # call Animal.__init__
        self.breed = breed
        print(f"Dog init: {self.breed}")

d = Dog("Buddy", "Labrador")
print("Name :", d.name)
print("Breed:", d.breed)
````

````python
class Vehicle:
	def __init__(self, brand):
		self.brand = brand
	
class EV(Vehicle):
	def __init__(self, brand, battery_cap):
		super().__init__(brand)
		self.battery_cap = battery_cap
	
	def print_details(self):
		print(f"{self.brand}:{self.battery_cap}")

class fuel_car(Vehicle):
    def __init__(self, brand, fuel_type, model):
        super().__init__(brand)  # Call the parent class constructor
        self.fuel_type = fuel_type
        self.model = model # Initialize the model attribute

    def print_details(self):
        print(f"{self.brand}:{self.fuel_type}:{self.model}")
		
car = fuel_car("Toyota", "Petrol", "Corolla")
car1 = EV("Tesla", "100 kWh")
car.print_details()
car1.print_details()
````

# 5. Polymorphism

**Polymorphism** means *many forms* - the **same method name** behaves differently for different classes.

### Forms of Polymorphism in Python

| Form | Description |
|------|-------------|
| **Method overriding** | Child class redefines a parent's method |
| **Duck typing** | If it looks like a duck and quacks like a duck... (any object with the right method works) |
| **Operator overloading** | Make `+`, `-`, `==` etc. work on custom objects |

````python
# Polymorphism via method overriding
class Animal:
    def sound(self):
        print("Some generic sound")

class Dog(Animal):
    def sound(self):
        print("Woof!")

class Cat(Animal):
    def sound(self):
        print("Meow!")

class Cow(Animal):
    def sound(self):
        print("Moo!")

# Same method call - different behavior
for animal in [Dog(), Cat(), Cow()]:
    animal.sound()
````

````python
# Duck typing - no inheritance needed
class Duck:
    def quack(self):
        print("Quack!")

class Person:
    def quack(self):
        print("I'm pretending to be a duck!")

def make_it_quack(thing):
    thing.quack()                # works for ANY object that has quack()

make_it_quack(Duck())
make_it_quack(Person())
````

## 5.1 Operator Overloading

You can define how operators (`+`, `-`, `*`, `==`, etc.) behave for your custom objects using **dunder (double underscore) methods**.

| Operator | Dunder Method |
|----------|---------------|
| `+` | `__add__` |
| `-` | `__sub__` |
| `*` | `__mul__` |
| `/` | `__truediv__` |
| `==` | `__eq__` |
| `<` | `__lt__` |
| `>` | `__gt__` |
| `str(obj)` | `__str__` |
| `len(obj)` | `__len__` |

````python
# Operator overloading example
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):                    # p1 + p2
        return Point(self.x + other.x, self.y + other.y)

    def __sub__(self, other):                    # p1 - p2
        return Point(self.x - other.x, self.y - other.y)

    def __eq__(self, other):                     # p1 == p2
        return self.x == other.x and self.y == other.y

    def __str__(self):                           # str(p1) / print(p1)
        return f"Point({self.x}, {self.y})"

p1 = Point(2, 3)
p2 = Point(4, 5)

print("p1     :", p1)
print("p2     :", p2)
print("p1 + p2:", p1 + p2)
print("p2 - p1:", p2 - p1)
print("p1 ==  :", p1 == Point(2, 3))
````

# 6. Method (Function) Overriding

**Method overriding** happens when a **child class** redefines a method that exists in the **parent class**. The child's version replaces the parent's version when called on a child object.

### Key Points
- Same method name
- Same parameters
- Different (child-specific) behavior
- Can still call the parent's version using `super()`

````python
# Method overriding
class Vehicle:
    def start(self):
        print("Vehicle is starting...")

class Car(Vehicle):
    def start(self):                           # OVERRIDES parent's start()
        print("Car is starting with a key.")

class ElectricCar(Vehicle):
    def start(self):
        print("Electric car is starting silently.")

Vehicle().start()
Car().start()
ElectricCar().start()
````

````python
# Overriding while still calling the parent's logic
class Employee:
    def show(self):
        print("I am an Employee.")

class Manager(Employee):
    def show(self):
        super().show()                     # call parent's version first
        print("I am also a Manager.")

m = Manager()
m.show()
````

# 7. Encapsulation

**Encapsulation** means **bundling data and methods together** AND **restricting direct access** to some of the object's internals. It protects an object's state from accidental changes.

### Access Modifiers in Python

| Type | Naming | Meaning |
|------|--------|---------|
| **Public** | `name` | Accessible from anywhere |
| **Protected** | `_name` | Convention: "internal use". Still accessible, but you *shouldn't* touch it from outside |
| **Private** | `__name` | Name-mangled to `_ClassName__name`. Hardest to access from outside |

Python doesn't strictly enforce private/protected - it relies on **convention** (underscores) and **name mangling**.

````python
# Public, Protected, Private attributes
class Account:
    def __init__(self, owner, balance):
        self.owner = owner          # public
        self._bank = "ABC Bank"     # protected (convention)
        self.__balance = balance    # private (name-mangled)

    def show(self):
        print(f"Owner   : {self.owner}")
        print(f"Bank    : {self._bank}")
        print(f"Balance : {self.__balance}")

a = Account("Alice", 1000)
a.show()

# Public - works
print("\nPublic owner:", a.owner)

# Protected - works but discouraged
print("Protected bank:", a._bank)

# Private - direct access fails
try:
    print(a.__balance)
except AttributeError as e:
    print("Error:", e)

# Name-mangled access (NOT recommended in real code)
print("Mangled :", a._Account__balance)
````

````python
# Encapsulation in action - controlled access via methods (getters/setters)
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.__balance = balance     # private

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited {amount}. New balance: {self.__balance}")
        else:
            print("Deposit must be positive.")

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            print(f"Withdrew {amount}. New balance: {self.__balance}")
        else:
            print("Invalid amount or insufficient funds.")

    def get_balance(self):           # getter
        return self.__balance

acc = BankAccount("Miraj", 500)
acc.deposit(200)
acc.withdraw(100)
acc.withdraw(5000)                   # rejected
print("Final balance:", acc.get_balance())
````

## Exercise 1: Rectangle


Create a `Rectangle` class with `width` and `height`. Add methods `area()` and `perimeter()`.

````python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * (self.width + self.height)

r = Rectangle(5, 3)
print("Area      :", r.area())
print("Perimeter :", r.perimeter())
````

## Exercise 2: Student Grades

Create a `Student` class that stores name, marks (a list) and has methods for average and grade.

````python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks               # list of marks

    def average(self):
        return sum(self.marks) / len(self.marks)

    def grade(self):
        avg = self.average()
        if avg >= 90: return "A"
        if avg >= 75: return "B"
        if avg >= 60: return "C"
        if avg >= 40: return "D"
        return "F"

s = Student("Miraj", [85, 92, 78, 88])
print("Name    :", s.name)
print("Average :", s.average())
print("Grade   :", s.grade())
````

## Exercise 3: Circle with Operator Overloading

Build a `Circle` class. Overload `+` (add radii), `==` (equal radii) and `<` (compare radius).

````python
import math

class Circle:
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2

    def __add__(self, other):
        return Circle(self.radius + other.radius)

    def __eq__(self, other):
        return self.radius == other.radius

    def __lt__(self, other):
        return self.radius < other.radius

    def __str__(self):
        return f"Circle(r={self.radius})"

c1 = Circle(5)
c2 = Circle(3)
c3 = c1 + c2

print("c1     :", c1, "area =", round(c1.area(), 2))
print("c2     :", c2)
print("c1+c2  :", c3)
print("c1==c2 :", c1 == c2)
print("c2<c1  :", c2 < c1)
````

## Exercise 4: Library / Book - Inheritance

Build a `Book` parent class and `EBook` child class that adds a `file_size` attribute.

````python
class Book:
    def __init__(self, title, author, price):
        self.title = title
        self.author = author
        self.price = price

    def info(self):
        print(f"{self.title} by {self.author} - ${self.price}")

class EBook(Book):
    def __init__(self, title, author, price, file_size):
        super().__init__(title, author, price)
        self.file_size = file_size

    def info(self):
        super().info()
        print(f"  File size: {self.file_size} MB")

b = Book("Clean Code", "Robert Martin", 30)
e = EBook("Python 101", "Mike Driscoll", 15, 5)

b.info()
e.info()
````

## Exercise 5: Shape Hierarchy - Polymorphism

Build a `Shape` base class and `Square`, `Triangle`, `Circle` subclasses that all override `area()`.

````python
import math

class Shape:
    def area(self):
        return 0

class Square(Shape):
    def __init__(self, side):
        self.side = side
    def area(self):
        return self.side ** 2

class Triangle(Shape):
    def __init__(self, base, height):
        self.base = base
        self.height = height
    def area(self):
        return 0.5 * self.base * self.height

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    def area(self):
        return math.pi * self.radius ** 2

shapes = [Square(4), Triangle(5, 6), Circle(3)]
for s in shapes:
    print(f"{type(s).__name__:<10} area = {round(s.area(), 2)}")
````

## Exercise 6: Bank Account - Encapsulation

Build a `BankAccount` class with a **private** balance, plus deposit / withdraw / get_balance methods.

````python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.__balance = balance         # private

    def deposit(self, amount):
        if amount <= 0:
            return "Amount must be positive"
        self.__balance += amount
        return f"Deposited {amount}. Balance: {self.__balance}"

    def withdraw(self, amount):
        if amount <= 0:
            return "Amount must be positive"
        if amount > self.__balance:
            return "Insufficient funds"
        self.__balance -= amount
        return f"Withdrew {amount}. Balance: {self.__balance}"

    def get_balance(self):
        return self.__balance

acc = BankAccount("Alice", 1000)
print(acc.deposit(500))
print(acc.withdraw(200))
print(acc.withdraw(5000))
print("Final :", acc.get_balance())
````
