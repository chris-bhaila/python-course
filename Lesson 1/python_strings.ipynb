# Python Strings - Methods, Functions & Slicing

A **string** in Python is a sequence of characters enclosed inside **single quotes `' '`**, **double quotes `" "`**, or **triple quotes `''' '''` / `""" """`** (used for multi-line strings).

### Key Characteristics

| Feature | Description |
|---------|-------------|
| **Immutable** | Once created, a string cannot be changed |
| **Ordered** | Characters have a definite position (index) |
| **Iterable** | Can be looped character-by-character |
| **Indexed** | Supports positive (`0, 1, 2...`) and negative (`-1, -2...`) indexing |
| **Sliceable** | Supports slicing to extract sub-strings |

## 1. Creating Strings

Strings can be created using single, double, or triple quotes.

````python
# Different ways to create strings
s1 = 'Hello'                  # Single quotes
s2 = "World"                  # Double quotes
s3 = '''This is a
multi-line string'''          # Triple quotes
s4 = """Also a
multi-line string"""

print("s1:", s1)
print("s2:", s2)
print("s3:", s3)
print("s4:", s4)
print("Type:", type(s1))
````

## 2. String Indexing

Every character in a string has an index:
- **Positive indexing** starts from `0` (left to right)
- **Negative indexing** starts from `-1` (right to left)

```
 P   Y   T   H   O   N
 0   1   2   3   4   5    <-- positive index
-6  -5  -4  -3  -2  -1    <-- negative index
```

````python
text = "PYTHON"

print("First character  :", text[0])    # P
print("Third character  :", text[2])    # T
print("Last character   :", text[-1])   # N
print("Second last char :", text[-2])   # O
````

## 3. String Slicing

Slicing extracts a **substring** using the syntax:

```python
string[start : stop : step]
```

| Parameter | Description | Default |
|-----------|-------------|---------|
| **start** | Index where slicing begins (inclusive) | `0` |
| **stop**  | Index where slicing ends (**exclusive**) | end of string |
| **step**  | Increment between characters | `1` |

> **Note:** The character at index `stop` is **not included** in the result.

````python
text = "Hello, Python!"

# Basic slicing
print("text[0:5]   :", text[0:5])     # Hello
print("text[7:13]  :", text[7:13])    # Python
print("text[:5]    :", text[:5])      # Hello  (from start)
print("text[-7:]    :", text[-7:])      # Python! (till end)
print("text[:]     :", text[:])       # Full copy of the string
````

## 4. Popular String Methods

Python provides many **built-in methods** to work with strings. Since strings are **immutable**, every method returns a **new string** (the original is not modified).

### Quick Reference Table

| Category | Methods |
|----------|---------|
| **Case Conversion** | `upper()`, `lower()`, `title()`, `capitalize()` |
| **Whitespace / Trim** | `strip()`, `lstrip()`, `rstrip()` |
| **Search / Check** | `find()`, `rfind()`, `index()`, `count()`, `startswith()`, `endswith()`, `in` operator |
| **Modify** | `replace()`, `split()`, `splitlines()`, `join()` |
| **Alignment / Padding** | `center()`, `ljust()`, `rjust()` |
| **Type Checking** | `isalpha()`, `isdigit()`, `isalnum()`, `isspace()`, `islower()`, `isupper()`, `istitle()` |
| **Encoding** | `encode()`, `decode()` |

### 4.1 Case Conversion Methods

````python
text = "hello python WORLD"

print("upper()     :", text.upper())        # HELLO PYTHON WORLD
print("lower()     :", text.lower())        # hello python world
print("title()     :", text.title())        # Hello Python World
print("capitalize():", text.capitalize())   # Hello python world
````

### 4.2 Whitespace / Trim Methods

Used to remove unwanted spaces or characters from a string.

````python
text = "   Hello Python   "

print("Original :", repr(text))
print("strip()  :", repr(text.strip()))    # Both sides
print("lstrip() :", repr(text.lstrip()))   # Left side
print("rstrip() :", repr(text.rstrip()))   # Right side

# strip() can also remove specific characters
url = "###python###"
print(url.strip('#'))       # python
````

### 4.3 Search & Check Methods

````python
text = "Learn Python programming with Python"

# find() -> returns index of first occurrence, -1 if not found
print("find('Python')  :", text.find("Python"))     # 6
print("find('Java')    :", text.find("Java"))       # -1

# rfind() -> last occurrence
print("rfind('Python') :", text.rfind("Python"))    # 30

# index() -> like find() but raises ValueError if not found
print("index('Python') :", text.index("Python"))    # 6

# count() -> number of occurrences
print("count('Python') :", text.count("Python"))    # 2

# startswith() / endswith()
print("startswith('Learn'):", text.startswith("Learn"))    # True
print("endswith('Python') :", text.endswith("Python"))     # True

# 'in' operator - membership test
print("'Python' in text   :", "Python" in text)            # True
````

### 4.4 Modify Methods - replace, split, join

````python
# replace(old, new) - Replaces all occurrences
text = "I like Java. Java is great."
print("replace():", text.replace("Java", "Python"))

# replace(old, new, count) - Replace only 'count' occurrences
print("replace() with count:", text.replace("Java", "Python", 1))
````

````python
# split() - splits the string into a list based on a separator (default is whitespace)
sentence = "Python is a great language"
print("split()       :", sentence.split())          # By whitespace

fruits= "apple|banana|cherry|mango"
print("split('|')    :", fruits.split("|"))            # By pipe

# splitlines() - splits by newline characters
para = "Line 1\nLine 2\nLine 3"
print("splitlines() :", para.splitlines())
````

````python
# join() - joins a list of strings into one string
words = ["Python", "is", "awesome"]
print("Join with space:", " ".join(words))       # Python is awesome
print("Join with '-'  :", "-".join(words))       # Python-is-awesome
print("Join with ''   :", "".join(words))        # Pythonisawesome
````

### 4.5 Alignment & Padding Methods

````python
text = "Python"

print("center(20, '*') :", text.center(20, '*'))   # *******Python*******
print("ljust(15, '-')  :", text.ljust(15, '-'))    # Python---------
print("rjust(15, '-')  :", text.rjust(15, '-'))    # ---------Python
````

### 4.6 Type Checking Methods

These methods return `True` or `False`.

````python
print("'Python'.isalpha() :", "Python".isalpha())      # True (only letters)
print("'12345'.isdigit()  :", "12345".isdigit())       # True (only digits)
print("'Python3'.isalnum():", "Python3".isalnum())     # True (letters + digits)
print("'   '.isspace()    :", "   ".isspace())         # True (only whitespace)
print("'hello'.islower()  :", "hello".islower())       # True (all lowercase)
print("'HELLO'.isupper()  :", "HELLO".isupper())       # True (all uppercase)
print("'Hello World'.istitle():", "Hello World".istitle())  # True (title-cased)
````

## 5. Built-in Functions Applicable to Strings

Python provides several **built-in functions** (not methods) that work with strings.

| Function | Purpose |
|----------|---------|
| `len(s)`      | Returns number of characters |
| `str(x)`      | Converts any object to string |
| `type(s)`     | Returns the type |
| `min(s)` / `max(s)` | Returns smallest / largest character |
| `sorted(s)`   | Returns a sorted list of characters |
| `reversed(s)` | Returns a reversed iterator |
| `enumerate(s)`| Returns index-character pairs |
| `list(s)`     | Converts a string into a list of characters |

````python
text = "Python"

print("len(text)      :", len(text))         # 6
print("str(123)       :", str(123))          # '123'
print("type(text)     :", type(text))        # <class 'str'>
````

````python
text = "python"

print("min(text)      :", min(text))         # h 
print("max(text)      :", max(text))         # y 
print("sorted(text)   :", sorted(text))      # ['h', 'n', 'o', 'p', 't', 'y']
print("list(text)     :", list(text))        # ['p', 'y', 't', 'h', 'o', 'n']
print("reversed(text) :", ''.join(reversed(text)))   # nohtyp
````
