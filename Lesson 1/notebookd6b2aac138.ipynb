## Is Python Compiled or Interpreted?

Python is both **compiled** and **interpreted**.

### Step 1: Compilation

Python converts source code (`.py`) into **bytecode** (`.pyc`).

### Step 2: Interpretation

The Python Virtual Machine (PVM) interprets and executes the bytecode.

### Simple Summary

```text
Python Code (.py)
        ↓
   Bytecode (.pyc)
        ↓
Python Virtual Machine (PVM)
        ↓
      Output
```

### Interview Answer

Python is both compiled and interpreted. It first compiles the source code into bytecode and then the Python Virtual Machine (PVM) interprets and executes that bytecode.

## Rules for Python Variables

- A variable name must start with a letter or the underscore character (`_`)
- A variable name cannot start with a number
- A variable name can only contain alphanumeric characters and underscores (`A–Z`, `a–z`, `0–9`, `_`)
- Variable names are case-sensitive (`age`, `Age`, and `AGE` are three different variables)
- A variable name cannot be any of the Python keywords

## 2.2 Data Types

`Variables can store data of different types, and different types can do different things.`

### What is a Data Type?

A **data type** is a classification that tells Python:
- **What kind of value** a variable holds (text, number, true/false, etc.)
- **How much memory** to allocate for it
- **What operations** can be performed on it (e.g., you can add two numbers, but you can't divide two strings)

Python is a **dynamically typed** language, meaning you don't need to declare the type of a variable - Python figures it out automatically based on the value you assign.

```python
x = 5         # Python knows this is an int
x = "Hello"   # Now Python knows this is a str
```

### Built-in Data Types in Python

Python has several built-in data types, grouped into categories:

| Category | Data Types |
|----------|------------|
| **Text Type** | `str` |
| **Numeric Types** | `int`, `float`, `complex` |
| **Sequence Types** | `list`, `tuple`, `range` |
| **Mapping Type** | `dict` |
| **Set Types** | `set`, `frozenset` |
| **Boolean Type** | `bool` |
| **Binary Types** | `bytes`, `bytearray`, `memoryview` |
| **None Type** | `NoneType` |

> In this lesson, we focus on the **core beginner data types**: `str`, `int`, `float`, `complex`, `bool`, and `None`. The others (lists, tuples, dicts, sets) will be covered in upcoming lessons.

### The `type()` Function

Python provides a built-in function `type()` that returns the data type of any value or variable.

**Syntax:**
```python
type(value)
```

This is super useful when you want to check what kind of data you're working with.

### Quick Overview of the Core Data Types

| Data Type | Class Name | Example | Used For |
|-----------|------------|---------|----------|
| String | `str` | `"Hello"`, `'Python'` | Text, names, sentences |
| Integer | `int` | `10`, `-50`, `2026` | Whole numbers (counting) |
| Float | `float` | `3.14`, `-0.5`, `2.5e3` | Decimal numbers (measurements) |
| Complex | `complex` | `3 + 4j` | Scientific & engineering calculations |
| Boolean | `bool` | `True`, `False` | Yes/No, On/Off, conditions |
| NoneType | `NoneType` | `None` | Absence of value (empty) |

### How Python Assigns a Data Type

Python decides the data type based on the **value** you assign:

| If value is... | Data type will be... |
|----------------|----------------------|
| Enclosed in quotes `" "` or `' '` | `str` |
| A whole number (no decimal) | `int` |
| A number with a decimal point | `float` |
| A number with `j` at the end | `complex` |
| `True` or `False` | `bool` |
| `None` | `NoneType` |

Let's see this in action with the `type()` function in the next cell.

````python
# Using type() to check the data type of different values

print("Value\t\tData Type")
print("-" * 35)
print("'Hello'\t\t", type("Hello"))
print("100\t\t", type(100))
print("3.14\t\t", type(3.14))
print("3 + 4j\t\t", type(3 + 4j))
print("True\t\t", type(True))
print("None\t\t", type(None))
````

````python
# Checking data type of variables

name = "Alice"          # str
age = 25                # int
height = 5.6            # float
z = 2 + 3j              # complex
is_student = True       # bool
address = None          # NoneType

print("name     =", name, "->", type(name))
print("age      =", age, "->", type(age))
print("height   =", height, "->", type(height))
print("z        =", z, "->", type(z))
print("student? =", is_student, "->", type(is_student))
print("address  =", address, "->", type(address))
````

````python
# Python is DYNAMICALLY TYPED - the same variable can hold different types
x = 10
print("x =", x, ", type =", type(x))

x = "Now I'm a string"
print("x =", x, ", type =", type(x))

x = 3.14
print("x =", x, ", type =", type(x))

x = True
print("x =", x, ", type =", type(x))
````

### Why Data Types Matter

Different data types support different operations. Knowing the type helps you avoid errors.

| Operation | Works on | Example |
|-----------|----------|---------|
| `+` (addition) | numbers | `5 + 3 = 8` |
| `+` (concatenation) | strings | `"Hi" + " there" = "Hi there"` |
| `+` between str and int | NOT ALLOWED | `"Age: " + 25` -> Error! |

Mixing incompatible types causes a `TypeError`. See the example below.

````python
# Example of a TypeError - mixing string and integer with +
age_num = 25
age_str = "30"
sum_age = age_num + age_str  # This will raise a TypeError
````

## 4. format() Method, f-strings, & Escape Sequences

### 4.1 Old Way - String Concatenation

````python
name = "Alice"
age = 25

# Concatenation - tedious and error-prone
# message = "My name is " + name + " and I am " + age + " years old." # This will raise a TypeError because age is an int, not a str
message = "My name is " + name + " and I am " + str(age) + " years old."
print(message)
````

### 4.2 The `format()` Method

````python
name = "Alice"
age = 25

# Using {} as placeholders (positional)
print("My name is {} and I am {} years old.".format(name, age))

# Using indexes
print("{0} is {1}. Yes, {0}!".format(name, age))

# Using named arguments
print("My name is {var1} and I am {var2} years old.".format(var1=name, var2=age))
````

````python
pi = 3.141592653589793
# print("{:.2f}".format(pi))
# print("{:.2f}".format(67882.378987))

num1 = "{:.4f}".format(67882.378987)

print(num1)
````

````python
# Number formatting with format()
pi = 3.141592653589793

print("Pi is {:.2f}".format(pi))         # 2 decimal places
print("Pi is {:.4f}".format(pi))         # 4 decimal places

print("Big number: {:,}".format(1234567))  # with commas

print("Padded: '{:>10}'".format("Hi"))    # right-aligned
print("Padded: '{:<10}'".format("Hi"))    # left-aligned
print("Padded: '{:^10}'".format("Hi"))    # centered
````

### 4.3 f-strings (RECOMMENDED) - Python 3.6+

````python
name = "Alice"
age = 25

# f-string - prefix with 'f' and embed variables inside {}
print(f"My name is {name} and I am {age} years old.")

# Expressions inside f-strings
print(f"Next year, I will be {age + 1}.")
print(f"In 5 years, I will be {age + 5}.")

# Calling methods inside f-strings
print(f"My name in uppercase: {name.upper()}")
print(f"Length of name: {len(name)}")
````

````python
# f-string formatting
pi = 3.141592653589793
price = 1234.5678

print(f"Pi is {pi:.2f}")               # 3.14
print(f"Pi is {pi:.5f}")               # 3.14159

print(f"Price: ${price:.2f}")           # $1234.57
print(f"With commas: {1234567:,}")      # 1,234,567
````

````python
# Creating a nice formatted table with f-strings
Name = "Alice"
Score = 95

Name1 = "Bob"
Score1 = 87

print(f"{'Name':<10} {'Score':>5}")
print("-" * 18)
print(f"{Name:<10} {Score:>5}")
print(f"{Name1:<10} {Score1:>5}")
````

### 4.4 Escape Sequences

````python
# \n - new line
print("Line 1\nLine 2\nLine 3")

print()  # blank line

# \t - tab
print("Name:\tAlice")
print("Age:\t25")
print("City:\tMumbai")
````

````python
# \' and \" - escaping quotes
print('It\'s a beautiful day!')
print("She said, \"Hello!\"")

# \\ - backslash
print("File path: C:\\Users\\Alice\\Documents")
````

````python
# Raw strings - use 'r' prefix to ignore escape sequences
print(r"C:\Users\Alice\new_folder")        # printed as-is
print(r"\n is a newline character")        # \n is NOT processed
````

## 5. Basic Arithmetic & Operators

### 5.2 Assignment Operators

````python
x = 10
print("Initial x:", x)

x += 5      # x = x + 5
print("After x += 5:", x)

x -= 3      # x = x - 3
print("After x -= 3:", x)

x *= 2      # x = x * 2
print("After x *= 2:", x)

x /= 4      # x = x / 4
print("After x /= 4:", x)

x **= 2     # x = x ** 2
print("After x **= 2:", x)
````

### 5.3 Comparison Operators (return True or False)

````python
a = 10
b = 5

print(f"a == b: {a == b}")    # equal to
print(f"a != b: {a != b}")    # not equal to
print(f"a > b:  {a > b}")     # greater than
print(f"a < b:  {a < b}")     # less than
print(f"a >= b: {a >= b}")    # greater than or equal
print(f"a <= b: {a <= b}")    # less than or equal
````

### 5.4 Logical Operators

### AND (`and`)

| A | B | A and B |
|---|---|----------|
| True | True | True |
| True | False | False |
| False | True | False |
| False | False | False |

### OR (`or`)

| A | B | A or B |
|---|---|---------|
| True | True | True |
| True | False | True |
| False | True | True |
| False | False | False |

### NOT (`not`)

| A | not A |
|---|-------|
| True | False |
| False | True |

````python
age = 20
has_license = True
has_car = False

# 'and' - True if BOTH are True
print("Can drive own car:", age >= 18 and has_license and has_car)

# 'or' - True if AT LEAST ONE is True
print("Can use transportation:", has_car or has_license)

# 'not' - inverts the value
print("Doesn't have a car:", not has_car)

# Combining them
is_eligible = (age >= 18) and (has_license or has_car)
print("Is eligible:", is_eligible)
````

````python
print(f"Can drive own car: {age >= 18 and has_license and has_car}")
````

### 5.5 Operator Precedence (PEMDAS)

````python
# Without parentheses - multiplication before addition
print(2 + 3 * 4)        # 14, not 20

# With parentheses - addition first
print((2 + 3) * 4)      # 20

# Exponent has highest precedence (after parentheses)
print(2 ** 3 + 1)       # 9 (8 + 1)

# Tricky one - ** is right-associative
print(2 ** 3 ** 2)      # 512 (2 ** 9), NOT 64
# 3**2  -> 9
# 2**9  -> 512

# Complex expression
result = (10 + 5) * 2 - 8 / 4 + 3 ** 2
print(f"Result: {result}")  # 37.0
````

### Expression

```python
result = (10 + 5) * 2 - 8 / 4 + 3 ** 2
```

5. Subtraction `-`

```python

```

Expression becomes:

```python

```

6. Addition `+`


## 6. Type Casting, Type Checking & Validation

### 6.1 Checking Types with `type()`

````python
# Check the type of any value
print(type(42))            # int
print(type(3.14))          # float
print(type("hello"))       # str
print(type(True))          # bool
print(type(None))          # NoneType
print(type(3 + 4j))        # complex

# Check the type of variables
name = "Alice"
age = 25
height = 5.6

print(f"\nname is of type: {type(name)}")
print(f"age is of type:  {type(age)}")
print(f"height is of type: {type(height)}")
````

### 6.2 Type Casting (Converting Between Types)

````python
# String to Integer
num_str = "42"
num_int = int(num_str)
print(f"'42' as int: {num_int}, type: {type(num_int)}")
print(f"Adding 8: {num_int + 8}")
````

````python
# String to Float
price_str = "19.99"
price = float(price_str)
print(f"Price: {price}, type: {type(price)}")
print(f"Double the price: {price * 2}")
````

````python
# Number to String
age = 25
age_str = str(age)
print(f"Type after str(): {type(age_str)}")

message = "I am " + age_str + " years old."
print(message)
````

````python
# Float to Integer - TRUNCATES (cuts off decimals), doesn't round!
print(int(3.99))      # 3, not 4
print(int(3.01))      # 3
print(int(-3.99))     # -3, not -4
````

````python
# To round properly, use round()
print(round(3.99))    # 4
print(round(3.01))    # 3
print(round(3.5))     # 4
````

````python
# Integer to Float
x = 5
y = float(x)
print(f"{x} -> {y}")
print(f"Type: {type(y)}")
````

````python
# Converting to Boolean
# 0, empty values, and None are falsy. Everything else is truthy.

print("bool(0):    ", bool(0))         # False
print("bool(1):    ", bool(1))         # True
print("bool(-5):   ", bool(-5))        # True (any non-zero number)
print("bool(''):   ", bool(""))        # False (empty string)
print("bool('hi'): ", bool("hi"))      # True
print("bool([]):   ", bool([]))        # False (empty list)
print("bool([0]):  ", bool([0]))       # True (non-empty list)
print("bool(None): ", bool(None))      # False
````

### 6.3 Type Checking with `isinstance()` (Preferred)

````python
age = 25
name = "Alice"
price = 99.99

# Check single type
print(isinstance(age, int))            # True
print(isinstance(name, str))           # True
print(isinstance(age, str))            # False
````

````python
# Check multiple types at once - use a tuple
print(isinstance(price, (int, float))) # True (price is a float)
print(isinstance(age, (int, float)))   # True (age is an int)
print(isinstance(name, (int, float)))  # False
````
