# 1. Defining Functions with `def`

## 1.1 What is a Function?

A **function** is a reusable block of code that performs a specific task. Functions help you:

- **Avoid repetition** - write code once, use many times (DRY: Don't Repeat Yourself)
- **Organize code** - break a big program into smaller logical pieces
- **Improve readability** - a well-named function reads like a sentence
- **Make testing easier** - test small pieces in isolation

### Syntax

```python
def function_name(parameters):
    """Optional docstring describing what the function does."""
    # body of the function
    return value         # optional
```

| Part | Meaning |
|------|---------|
| `def` | Keyword that tells Python you are defining a function |
| `function_name` | The name you give your function (snake_case is preferred) |
| `parameters` | Inputs the function accepts (optional) |
| `:` | Marks the start of the function body |
| Indented block | The code that runs when the function is called |
| `return` | Sends a value back to the caller (optional) |

````python
def save_student_details(name, address, phone:int)->None: #(parameters)
    print(f"Name: {name}\nAddress: {address}\nPhone: {phone}\n")

save_student_details("Chris", "Bhaktapur", 977)
save_student_details("Chris", "Bhaktapur", "978")
````

````python
# A simple function with NO parameters and NO return value
def greet():
    print("Hello, welcome to Python Functions!")

# Calling the function
greet()
greet()        # can be called as many times as we want
````

````python
# A function with a docstring (the first string inside the body)
def say_hi(name):
    """Print a friendly greeting for the given name."""  #Doc-string
    print(f"Hi, {name}!")

say_hi("Alice")
say_hi("Bob")

# The docstring can be accessed via __doc__
print("Docstring:", say_hi.__doc__)
````

````python
# Functions are objects in Python
def add(a, b):
    return a + b

print("Function    :", add)
print("Type        :", type(add))
print("Name        :", add.__name__)

# Assigning a function to another name
sum_two = add
print("Same result :", sum_two(3, 4))
````

# 2. Parameters, Arguments & Return Statements

## 2.1 Parameters vs Arguments

These two words are often confused, but they mean different things:

| Term | Meaning | Example |
|------|---------|---------|
| **Parameter** | Variable listed in the function **definition** | `def greet(name):` -> `name` |
| **Argument** | Actual value passed in the function **call** | `greet("Alice")` -> `"Alice"` |

## 2.2 The `return` Statement

- `return` sends a value back to whoever called the function.
- A function **exits immediately** when it hits a `return`.
- If you omit `return`, the function returns `None` automatically.

````python
# Function with parameters and a return value
def add(a, b):                # a, b are PARAMETERS
    return a + b

result = add(5, 3)            # 5, 3 are ARGUMENTS
print("5 + 3 =", result)

# The returned value can be used in any expression
print("Double :", add(10, 20) * 2)
````

````python
# Difference between print() and return
def square_print(n):
    print(n * n)              # only displays - returns None

def square_return(n):
    return n * n              # gives the value back

a = square_print(5)           # prints 25, but a = None
b = square_return(5)          # b = 25

print("a =", a)
print("b =", b)
print("b + 10 =", b + 10)     # we can use b in further calculations
````

````python
# return exits the function immediately
def check_sign(n):
    if n > 0:
        return "positive"
    if n < 0:
        return "negative"
    return "zero"
    print("This line never runs!")   # unreachable code

print(check_sign(10))
print(check_sign(-5))
print(check_sign(0))
````

````python
# A function without return -> returns None
def show_message():
    print("Hello!")

result = show_message()
print("Returned :", result)            # None
````

# 3. Returning Multiple Values

Python lets you return **more than one value** from a function. Under the hood, Python actually packs them into a **tuple** and you can unpack them on the receiving side.

### Why is this useful?

- Get several related values in a single call (e.g., min and max).
- Avoid having to define a custom class for simple groupings.
- Combine perfectly with **tuple unpacking** for clean code.

````python
# Returning multiple values (actually a tuple)
def min_max(numbers):
    return min(numbers), max(numbers)

result = min_max([3, 7, 1, 9, 4])
print("Result :", result)
print("Type   :", type(result))         # tuple
````

````python
# Unpacking the returned values into separate variables
def min_max(numbers):
    return min(numbers), max(numbers)

low, high = min_max([3, 7, 1, 9, 4])
print("Low  :", low)
print("High :", high)
````

````python
# Returning 3+ values
def stats(numbers):
    total = sum(numbers)
    count = len(numbers)
    avg = total / count
    return total, count, avg

t, c, a = stats([10, 20, 30, 40, 50])
print(f"Total = {t}, Count = {c}, Average = {a}")
````

````python
# Returning different data types together
def parse_user(text):
    name, age = text.split(",")
    return name.strip(), int(age)

name, age = parse_user("Alice, 25")
print("Name :", name, type(name))
print("Age  :", age,  type(age))
````

````python
# You can also return a list or dict if it suits the data better
def user_info(name, age, city):
    return {"name": name, "age": age, "city": city}

info = user_info("Alice", 25, "Mumbai")
print(info)
print("City:", info["city"])
````

# 4. Default & Keyword Arguments

## 4.1 Default Arguments

You can give a parameter a **default value**. If the caller does not pass an argument for it, the default is used.

```python
def greet(name="Guest"):
    print("Hello,", name)
```

**Rule:** Parameters with defaults must come **after** parameters without defaults.

## 4.2 Keyword Arguments

You can pass arguments by **name** at call time. This makes the call more readable and the order doesn't matter.

```python
create_user(name="Alice", age=25)
create_user(age=25, name="Alice")   # same result
```

## 4.3 `*args` and `**kwargs`

| Syntax | Captures | Available as |
|--------|----------|--------------|
| `*args` | Extra **positional** arguments | A tuple |
| `**kwargs` | Extra **keyword** arguments | A dictionary |

````python
# Default argument
def greet(name="Guest"):
    print(f"Hello, {name}!")

greet()                 # uses default -> Guest
greet("Alice")          # overrides default
````

````python
# Multiple defaults
def create_user(name, age=18, country="India"):
    print(f"User: {name}, Age: {age}, Country: {country}")

create_user("Alice")
create_user("Bob", 25)
create_user("Charlie", 30, "USA")
````

````python
# Keyword arguments - order doesn't matter, very readable
def create_user(name, age, country):
    print(f"User: {name}, Age: {age}, Country: {country}")

create_user(name="Alice", age=25, country="India")
create_user(country="USA", name="Bob", age=30)   # different order - same result

# Mixing positional and keyword arguments
# Rule: positional first, then keyword
create_user("Charlie", country="UK", age=22)
````

````python
# *args - collect extra POSITIONAL arguments into a tuple
def total(*numbers):
    print("Got      :", numbers, type(numbers))
    return sum(numbers)

print("Sum =", total(1, 2, 3))
print("Sum =", total(10, 20, 30, 40, 50))
````

````python
# **kwargs - collect extra KEYWORD arguments into a dict
def show_profile(**info):
    print("Got :", info, type(info))
    for key, value in info.items():
        print(f"  {key} -> {value}")

show_profile(name="Alice", age=25, city="Mumbai")
````

````python
def my_bio(name, address, bloodgroup, age):
    print(f"{name}, {address}, {bloodgroup}, {age}")

my_bio("Chris", "BKT", "B+ve", 22)

# In real-world scenario, a bio form will have multiple questions, so **kwargs is the way to go, making the user themself put all the
# necessary information

def bio(**details):
    print(f"Details:    {details}", type(details))

bio(name="Chris", address="BKT", faculty="BCA", age="22")
````

````python
# Combining everything in one function
# Order: regular -> *args -> default -> **kwargs
def describe(title, *tags, author="Unknown", **meta):
    print("Title :", title)
    print("Tags  :", tags)
    print("Author:", author)
    print("Meta  :", meta)

describe("Python Notes",
         "beginner", "reference",
         author="Alice",
         pages=120, language="English")
````

# 5. Anonymous Functions (`lambda`)

A **lambda** is a small, **one-line, unnamed function**. It is handy when you need a quick function for a short time (often as an argument to another function).

### Syntax

```python
lambda arguments: expression
```

| Feature | `def` function | `lambda` function |
|---------|---------------|-------------------|
| Name | Has a name | Anonymous |
| Size | Multiple lines | Single expression |
| Statements | Allowed | NOT allowed (only an expression) |
| Return | Uses `return` | Implicit return |
| Best for | Reusable logic | Short, throwaway functions |

Lambdas are commonly used with `map()`, `filter()`, `sorted()`, and `reduce()`.

````python
# Equivalent function written two ways
def square_def(x):
    return x * x

square_lambda = lambda x: x * x

print("def    :", square_def(5))
print("lambda :", square_lambda(5))
````

````python
# Lambda with multiple arguments
add = lambda a, b: a + b
is_even = lambda n: n % 2 == 0

print("Add     :", add(3, 7))
print("Is even :", is_even(10))
print("Is even :", is_even(7))
````

````python
# Lambda with map() - apply a function to every item
numbers = [1, 2, 3, 4, 5]

squares = list(map(lambda x: x ** 2, numbers))
print("Squares:", squares)

# Lambda with filter() - keep items where the lambda is True
evens = list(filter(lambda x: x % 2 == 0, numbers))
print("Evens  :", evens)
````

````python
salary = [10000, 20000, 30000, 40000, 50000]
new_salary = list(map(lambda x: x+5000, salary))
# new_salary
````

````python
[x+5000 for x in salary]
````

````python
# Lambda with sorted() - custom sort key
students = [("Alice", 85), ("Bob", 72), ("Charlie", 90)]

# Sort by score (the 2nd item of each tuple)
by_score = sorted(students, key=lambda s: s[1])
print("By score asc :", by_score)

# Sort by score descending
by_score_desc = sorted(students, key=lambda s: s[1], reverse=True)
print("By score desc:", by_score_desc)

# Sort by name length
by_name_len = sorted(students, key=lambda s: len(s[0]))
print("By name len  :", by_name_len)
````

````python
# Lambda with reduce() - combine items into a single value
from functools import reduce

numbers = [1, 2, 3, 4, 5]

product = reduce(lambda x, y: x * y, numbers)
print("Product:", product)            # 1*2*3*4*5 = 120

total = reduce(lambda x, y: x + y, numbers)
print("Total  :", total)              # 15
````

````python
# Conditional expression inside a lambda
grade = lambda score: "Pass" if score >= 40 else "Fail"

print(grade(75))
print(grade(30))
````

# 6. Scopes in Python: Local & Global

**Scope** = the region of code where a variable is accessible.

Python mainly uses two scopes that beginners care about:

| Scope | Description |
|-------|-------------|
| **Local** | Variables created inside a function. Only visible inside that function. |
| **Global** | Variables defined at the top level (outside any function). Visible everywhere. |

### Key Rules

- You can **read** a global variable from inside a function without any special keyword.
- To **modify** a global variable from inside a function, you must use the `global` keyword.
- If you assign to a name inside a function (without `global`), Python treats it as a **new local variable**.

### Keyword

| Keyword | Purpose |
|---------|---------|
| `global` | Modify a **global** variable from inside a function |

````python
# LOCAL scope - variable lives only inside the function
def my_func():
    x = 10                # x is local to my_func
    print("Inside :", x)

my_func()

# x doesn't exist out here
try:
    print(x)
except NameError as e:
    print("Error:", e)
````

````python
# GLOBAL scope - defined at the top level
x = 100                   # global variable

def show():
    print("Inside  :", x)    # READ is allowed without 'global'

show()
print("Outside :", x)
````

````python
# Trying to modify a global without 'global' creates a NEW local variable
x = 100

def update():
    x = 50                # this creates a new LOCAL x
    print("Inside :", x)

update()
print("Outside:", x)      # global x is still 100
````

````python
# Use the 'global' keyword to actually modify the global
x = 100

def update():
    global x
    x = 50                # now modifying the global x
    print("Inside :", x)

update()
print("Outside:", x)      # global x is now 50
````

````python
# Variable shadowing - same name in different scopes
name = "Global Alice"

def greet():
    name = "Local Bob"               # shadows the global 'name'
    print("Inside :", name)

greet()
print("Outside:", name)              # global is unchanged
````

## Exercise 1: Simple Calculator

Build a calculator that supports `+`, `-`, `*`, `/` using separate functions plus a `calculate()` dispatcher.

````python
# Simple Calculator
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        return "Error: cannot divide by zero"
    return a / b

def calculate(a, b, op):
    operations = {
        "+": add,
        "-": subtract,
        "*": multiply,
        "/": divide,
    }
    if op not in operations:
        return "Invalid operator"
    return operations[op](a, b)

print("10 + 5 =", calculate(10, 5, "+"))
print("10 - 5 =", calculate(10, 5, "-"))
print("10 * 5 =", calculate(10, 5, "*"))
print("10 / 5 =", calculate(10, 5, "/"))
print("10 / 0 =", calculate(10, 0, "/"))
print("10 ^ 5 =", calculate(10, 5, "^"))
````

## Exercise 3: Prime Number Checker

A **prime** number is greater than 1 and only divisible by 1 and itself.

````python
def is_prime(n):
    if n <= 1:
        return False
    # Only need to check up to the square root of n
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

for num in [1, 2, 3, 4, 5, 7, 11, 15, 17, 20, 29]:
    print(f"{num:>3} -> {'Prime' if is_prime(num) else 'Not prime'}")

# Print all primes up to 50
primes_under_50 = [n for n in range(2, 51) if is_prime(n)]
print("\nPrimes under 50:", primes_under_50)
````

## Exercise 4: Factorial

Factorial of n: `n! = n * (n-1) * (n-2) * ... * 1`

Written in two ways - using a loop and using recursion.

````python
# Iterative version (using a loop)
def factorial(n):
    if n < 0:
        return "Factorial not defined for negative numbers"
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

# Recursive version (function calling itself)
def factorial_recursive(n):
    if n <= 1:
        return 1
    return n * factorial_recursive(n - 1)

for n in [0, 1, 5, 7, 10]:
    print(f"{n}! = {factorial(n)}  (recursive: {factorial_recursive(n)})")
````

## Exercise 5: Fibonacci Series

Each number is the sum of the previous two: `0, 1, 1, 2, 3, 5, 8, 13, 21, ...`

````python
def fibonacci(n):
    """Return a list with the first n Fibonacci numbers."""
    series = []
    a, b = 0, 1
    for _ in range(n):
        series.append(a)
        a, b = b, a + b           # tuple unpacking to update both at once
    return series

print("First 10 :", fibonacci(10))
print("First 15 :", fibonacci(15))
````

## Exercise 6: Palindrome Checker

A **palindrome** reads the same forwards and backwards: `121`, `racecar`, `madam`.

````python
def is_palindrome(value):
    # Convert to string and ignore case + spaces
    text = str(value).lower().replace(" ", "")
    return text == text[::-1]      # compare with reversed version

tests = [121, 123, "racecar", "Hello", "madam", "Nurses run"]
for t in tests:
    print(f"{str(t):<12} -> {is_palindrome(t)}")
````

## Exercise 7: Count Vowels

````python
def count_vowels(text):
    vowels = "aeiouAEIOU"
    count = 0
    for ch in text:
        if ch in vowels:
            count += 1
    return count

# Pythonic one-liner version
def count_vowels_v2(text):
    return sum(1 for ch in text if ch.lower() in "aeiou")

samples = ["Hello World", "Python Programming", "AEIOU", "xyz"]
for s in samples:
    print(f"'{s}' -> {count_vowels(s)} vowels  (v2: {count_vowels_v2(s)})")
````

## Exercise 8: FizzBuzz

Classic interview question:
- Multiples of 3 -> print `Fizz`
- Multiples of 5 -> print `Buzz`
- Multiples of both 3 and 5 -> print `FizzBuzz`
- Otherwise -> print the number

````python
def fizzbuzz(n):
    for i in range(1, n + 1):
        if i % 15 == 0:
            print("FizzBuzz")
        elif i % 3 == 0:
            print("Fizz")
        elif i % 5 == 0:
            print("Buzz")
        else:
            print(i)

fizzbuzz(20)
````

## Exercise 9: Largest of Three Numbers

````python
def largest_of_three(a, b, c):
    if a >= b and a >= c:
        return a
    elif b >= a and b >= c:
        return b
    else:
        return c

# Shorter version using built-in max()
def largest_of_three_v2(a, b, c):
    return max(a, b, c)

print(largest_of_three(10, 25, 18))
print(largest_of_three(-5, -1, -10))
print(largest_of_three_v2(10, 25, 18))
````

## Exercise 10: Celsius <-> Fahrenheit Converter

Formulas:
- F = (C * 9/5) + 32
- C = (F - 32) * 5/9

````python
def celsius_to_fahrenheit(c):
    return (c * 9 / 5) + 32

def fahrenheit_to_celsius(f):
    return (f - 32) * 5 / 9

print("0 C  =", celsius_to_fahrenheit(0), "F")
print("100 C =", celsius_to_fahrenheit(100), "F")
print("37 C  =", celsius_to_fahrenheit(37), "F")

print("\n32 F  =", fahrenheit_to_celsius(32), "C")
print("212 F =", fahrenheit_to_celsius(212), "C")
print("98.6 F =", round(fahrenheit_to_celsius(98.6), 2), "C")
````

## Exercise 11: Password Strength Checker

A simple checker that rates a password as Weak, Medium, or Strong based on:
- Length
- Has uppercase letters
- Has digits
- Has special characters

````python
def password_strength(password):
    score = 0
    specials = "!@#$%^&*()-_+=<>?/"

    if len(password) >= 8:
        score += 1
    if any(ch.isupper() for ch in password):
        score += 1
    if any(ch.isdigit() for ch in password):
        score += 1
    if any(ch in specials for ch in password):
        score += 1

    if score <= 1:
        return "Weak"
    elif score <= 3:
        return "Medium"
    else:
        return "Strong"

passwords = ["abc", "password", "Password1", "Pass@123", "P@ssw0rd!2024"]
for p in passwords:
    print(f"{p:<15} -> {password_strength(p)}")
````
