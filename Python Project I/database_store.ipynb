# Store The Products into the DataBase

### 1) Loading the Json Data

```python
import json
json_path = "products.json"

# Load the JSON data
with open(json_path, 'r', encoding='utf-8') as f:
    data = json.load(f)
```

### 2) Inspecting the data

```python
for key, values in data.items():
    print(f"| Category: {key} | Number of products: {len(values)} |")
```

### 3) Make the database

```python
import sqlite3
db_name = "products.db"
conn = sqlite3.connect(db_name)
```

### 4) Make the Cursor

```python
cursor = conn.cursor() # this cursor object will be used to execute SQL commands
```

### 4) Make the Table

```python
products_table = """CREATE TABLE IF NOT EXISTS products (
                    id       INTEGER PRIMARY KEY AUTOINCREMENT,
                    name    TEXT    NOT NULL,
                    price    TEXT
                     )"""
                     
cursor.execute(products_table)
conn.commit()
```

### 5) Format the data 

```python
# Flatten the JSON into a list of tuples for bulk insert
rows = []
for category, products in data.items():
    for product in products:
        rows.append((
            category,
            product.get('name'),
            product.get('price'),
        ))
        
rows[0]
```

```python
rows
```

### 6) Insert Into DataBase

```python
insert_query = 'INSERT INTO products (name, price) VALUES (?, ?)'
clean_rows = [(name, price) for _, name, price in rows]
cursor.executemany(insert_query, clean_rows)
conn.commit()
```

### 7) View the data

```python
select_query = 'SELECT * FROM products LIMIT 2'
cursor.execute(select_query)
rows = cursor.fetchall()
rows
```

### 8) Update the product with id 1 price to MRP₹ 250,000

```python
update_query = 'UPDATE products SET price = ? WHERE id = ?'
cursor.execute(update_query, ('250,000', 1))
conn.commit()
```

### 9) View the updated Data

```python
select_query_for_specific_id = 'SELECT * FROM products WHERE id = ?'
cursor.execute(select_query_for_specific_id, (1,))
row = cursor.fetchone()
row
```

### 10) Delete the product having id = 2

```python
delete_query = 'DELETE FROM products WHERE id = ?'
cursor.execute(delete_query, (2,))
conn.commit()
```
