```python
from bs4 import BeautifulSoup
import requests

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/122.0.0.0 Safari/537.36",
    "Accept-Language": "en-US,en;q=0.9",
}

response = requests.get("https://yamahamusicstore.in/", headers=headers, timeout=10)

print(response.status_code)
```

```python
html_doc = response.text

soup = BeautifulSoup(html_doc, 'html.parser')
print(soup.prettify)
```

```python
categories_section = soup.find("div", class_="flat-animate-tab")
categories = categories_section.find_all("div", class_="catgcvr")
cat_list = {}
for cat in categories:
    cat_name = cat.get_text(strip=True)
    cat_link = cat.find("a").get("href")
    cat_list[cat_name] = cat_link

cat_list
```

```python
target_pages = 4

scrape = {}

for category, link in cat_list.items():
    for page in range(1, target_pages+1):
        dynamic_url = f"{link}?page={page}" #Dynamic url to continue scraping on other pages (dealing with pagination)

        html_doc = requests.get(dynamic_url, headers=headers, timeout=10) #Establish connection
        if html_doc.status_code != 200:
            break

        cat_soup = BeautifulSoup(html_doc.text, 'html.parser')

        try:
            product_container = cat_soup.find("div", class_="tf-list-layout wrapper-shop prodLising") #Select the container where products reside
            print(f"Scraping page {page} for category {category}")
        except Exception as e:
            print(f"No product card found in page {page} of category {category}")
            break

        try:
            product = product_container.find_all("div", class_="card-product-info") #Select each product inside the container
        except Exception as e:
            print(f"No products found in page {page} of category {category}")
            break

        for pro in product:
            product_name = pro.find("a").get_text(strip=True) #Select name of each product
            
            product_price = pro.find("span", class_="price").get_text(strip=True)

            if category not in scrape:
                scrape[category] = []
                
            scrape[category].append({
                "name": product_name,
                "price": product_price
            })
print("Scraping Complete!")
```
